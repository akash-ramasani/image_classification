```python
DATA_DIR = "./data"
```

```python
train_ds = torchvision.datasets.CIFAR10(root=DATA_DIR, train=True, download=True)
```

```python
test_ds = torchvision.datasets.CIFAR10(root=DATA_DIR, train=False, download=True)
```

```python
print("Train Dataset Size:", len(train_ds))
```

```python
print("Test Dataset Size:", len(test_ds))
```

```python
print("Train Dataset Classes:", train_ds.classes)
```

```python
print("Test Dataset Classes:", test_ds.classes)
```

```python
print("Train and Test Dataset has same classes:", train_ds.classes == test_ds.classes)
```

```python
plt.figure(figsize=(8, 6))
for i in range(12):
    img, label = train_ds[i]
    plt.subplot(3, 4, i + 1)
    plt.imshow(img)
    plt.title(train_ds.classes[label], fontsize=9)
    plt.axis("off")
plt.tight_layout()
plt.show()
```
